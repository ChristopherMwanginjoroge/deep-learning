```python
import numpy as np
import matplotlib.pyplot as plt


audio_data=example1['audio']

waveform = np.asarray(audio_data["array"])
sample_rate = audio_data["sampling_rate"]

if waveform.ndim != 1:
    raise ValueError(
        f"Expected mono audio (1D), but got shape {waveform.shape}. "
        "We should inspect the channel layout before plotting."
    )

time=np.arange(len(waveform))/sample_rate



plt.figure(figsize=(14, 4))

plt.plot(time, waveform)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Waveform of a Kenyan Swahili Recording")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

```

```python
target_rate=16000

original_rate=audio_data['sampling_rate']

if original_rate==target_rate:
  resampled_waveform = waveform
  print('Already in the correct sampling rate')

else:
  resampled_waveform =librosa.resample(
      waveform,
      orig_sr=original_rate,
      target_sr=target_rate
      )

```

```python
from huggingface_hub.hf_file_system import HfFileSystemResolvedPath
import librosa.display

audio_16k=np.asarray(resampled_waveform,dtype=np.float32)


if audio_16k.ndim != 1:
    raise ValueError(
        f"Expected mono audio (1D), but got shape {audio_16k.shape}. "
        "We should inspect the channel layout before plotting."
    )


sample_rate=16000

stft=librosa.stft(
    audio_16k,
    n_fft=400,
    hop_length=160
)

magnitude=np.abs(stft)

decibles=librosa.amplitude_to_db(magnitude,ref=np.max)

plt.figure(figsize=(14, 4))

librosa.display.specshow(
    decibles,
    sr=sample_rate,
    hop_length=160,
    x_axis='time',
    y_axis='hz',)

plt.title("Spectrogram of Kenyan Swahili Speech")
plt.show()

```
